```python
import numpy as np

def calc_impurity(X: np.ndarray, method: str) -> float:
    """
    Calculate the impurity of a dataset using the specified method.

    Parameters:
    X (array-like): The class labels for which to calculate impurity.
    method (str): The method to use for calculating impurity. Options are 'gini', 'entropy', or 'variance'.
        'variance' uses classification error: 1 - max_i p(i).

    Returns:
    float: The calculated impurity value.
    """
    if method == 'gini':
        # Calculate Gini impurity
        classes, counts = np.unique(X, return_counts=True)
        probabilities = counts / counts.sum()
        gini_impurity = 1 - np.sum(probabilities ** 2)
        return gini_impurity
    elif method == 'entropy':
        # Calculate entropy
        classes, counts = np.unique(X, return_counts=True)
        probabilities = counts / counts.sum()
        entropy = -np.sum(probabilities * np.log2(probabilities + 1e-9))  # Adding a small value to avoid log(0)
        return entropy
    elif method == 'rate':
        # Calculate classification error: 1 - max_i p(i)
        classes, counts = np.unique(X, return_counts=True)
        probabilities = counts / counts.sum()
        rate = 1 - np.max(probabilities)
        return rate
    else:
        raise ValueError("Invalid method specified. Choose from 'gini', 'entropy', or 'rate'.")
```
