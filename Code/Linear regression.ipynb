Left inverse for tall matrices

```python
import numpy as np

X = np.array([[45, 9], 
[50,10], [63,12], [70,8], [80,4]])
y = np.array([[6,5], 
[9,6], 
[8,9],
[3,2],
[2,4]])
print(X)
```

```python
Xb = np.concatenate((np.ones([len(X),1]),X), axis=1)
print(Xb)
```

```python
wb = np.linalg.inv(Xb.T@Xb)@Xb.T@y
print(wb)
# Show every feature coefficient for every target column.
for column in range(wb.shape[1]):
    terms = "".join(
        f" {wb[feature + 1, column]:+.5f}*x{feature + 1}"
        for feature in range(X.shape[1])
    )
    print(f"Regression formula for y-column {column + 1}: y{column + 1} = {wb[0, column]:.5f}{terms}")
```

Prediction

Enter raw feature values in `X_predict`, one sample per row, in the same column order as `X`. The code adds the bias column automatically. Each prediction column corresponds to the matching column of `y`.

```python
# EDIT INPUTS: raw features only; the leading bias value is added below.
X_predict = np.array([[63, 9]])
assert X_predict.ndim == 2 and X_predict.shape[1] == X.shape[1], "Use one row per sample and the same feature columns as X."
X_predict_b = np.concatenate((np.ones((len(X_predict), 1)), X_predict), axis=1)
y_pred = X_predict_b @ wb

print("Predicted y (rows = samples; columns = target columns):")
print(y_pred)
```

Mean squared error (MSE) for each y-column

`axis=0` averages over the sample rows and returns one MSE per target column.

```python
X_test = Xb
y_test = y

# include snippet above if testing data is same as training data

y_pred_mse = X_test @ wb
MSE = ((y_test - y_pred_mse)**2).mean(axis=0)
for column, mse in enumerate(MSE, start=1):
    print(f"MSE for y-column {column}: {mse:.8f}")
```
