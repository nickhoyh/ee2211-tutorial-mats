Left inverse for tall matrices

```python
import numpy as np

X = np.array([[1, 2],
               [0,6],
               [1,0],
               [0,5],
               [1,7]])
y = np.array([[1],
              [2],
              [3],
              [4],
              [5]])
print(X)
```

```python
Xb = np.concatenate((np.ones([len(X),1]),X), axis=1)
print(Xb)
```

```python
wb = np.linalg.inv(Xb.T@Xb)@Xb.T@y
print(wb)
print('Regression formula: y = {:.2f}*x + {:.2f}'.format(wb[1][0], wb[0][0]))
```

Prediction

```python
y_pred = np.array([[1,1,3]])@wb
print(y_pred)
```

Mean squared error (MSE) = np.mean((y - y_pred)**2)
